
Una aerolínea necesita desarrollar un sistema para gestionar los diferentes tipos de pasajeros y
calcular el costo de sus vuelos.

Todo pasajero tendrá: id, nombre, edad, equipaje_kg
- es_mayor_de_edad() 
- calcular_precio_vuelo()
- puede_abordar()
- obtener_informacion()

El id será inmutable, mientras que los demás datos podrán modificarse mediante getters y setters.
Los atributos deberán estar encapsulados.

#### Pasajero turista

El precio del vuelo se calcula de la siguiente manera:
- Precio base: $100.000
- Los primeros 15 kg de equipaje son gratuitos.
- Cada kg adicional cuesta $2.000.

El pasajero puede abordar siempre que: sea mayor de edad y  tenga como máximo 25 kg de equipaje.

#### Pasajero ejecutivo
- El precio base es: $200.000
- El pasajero ejecutivo tiene incluidos 30 kg de equipaje.
- Por cada kg adicional se cobran $3.000.

Además, los pasajeros ejecutivos tienen un 10% de descuento sobre el precio final si tienen más de 5 vuelos realizados.

Por lo tanto, esta clase tendrá un atributo adicional: vuelos_realizados

El pasajero puede abordar siempre que: sea mayor de edad y tenga como máximo 40 kg de equipaje.

#### Pasajero menor

- Precio base de $50.000.
- Puede llevar hasta 10 kg de equipaje sin costo.
- Cada kg adicional cuesta $1.500.

Para poder abordar debe: tener menos de 18 años y viajar acompañado por un adulto.

Por lo tanto tendrá un atributo: acompanado_por_adulto, que será booleano. 

#### Vuelo
Crear una clase Vuelo con: codigo, origen, destino, capacidad, lista de pasajeros
El código del vuelo será inmutable.
Debe tener los siguientes métodos:

- agregar_pasajero(pasajero)
- eliminar_pasajero(id_pasajero)
- hay_lugar()
- cantidad_pasajeros()
- calcular_recaudacion()

###### agregar_pasajero()
Debe agregar al pasajero solamente si:
    - Hay lugar disponible.
    - El pasajero puede abordar.

Debe devolver: True si fue agregado correctamente y False si no pudo ser agregado.

###### hay_lugar()
Debe devolver True si todavía hay lugares disponibles.

###### calcular_recaudacion()
Debe calcular cuánto dinero recaudó el vuelo utilizando:

pasajero.calcular_precio_vuelo()

No debe comprobar qué tipo de pasajero es.
