
Una aerolínea necesita desarrollar un sistema para gestionar los diferentes tipos de pasajeros y
calcular el costo de sus vuelos.

Todo pasajero tendrá: id, nombre, edad, equipaje_kg
- es_mayor_de_edad() 
- calcular_precio_vuelo()
- puede_abordar()
- obtener_informacion()

El id será inmutable, mientras que los demás datos podrán modificarse mediante getters y setters.
Los atributos deberán estar encapsulados.

#### Pasajero turista

El precio del vuelo se calcula de la siguiente manera:
- Precio base: $100.000
  
- Los primeros 15 kg de equipaje son gratuitos.

  
- Cada kg adicional cuesta $2.000.

El pasajero puede abordar siempre que: sea mayor de edad y  tenga como máximo 25 kg de equipaje.

#### Pasajero ejecutivo
- El precio base es: $200.000
  
- El pasajero ejecutivo tiene incluidos 30 kg de equipaje.
  
- Por cada kg adicional se cobran $3.000.

Además, los pasajeros ejecutivos tienen un 10% de descuento sobre el precio final si tienen más de 5 vuelos realizados.

Por lo tanto, esta clase tendrá un atributo adicional: vuelos_realizados

El pasajero puede abordar siempre que: sea mayor de edad y tenga como máximo 40 kg de equipaje.

#### Pasajero menor

- Precio base de $50.000.
  
- Puede llevar hasta 10 kg de equipaje sin costo.

  
- Cada kg adicional cuesta $1.500.

Para poder abordar debe: tener menos de 18 años y viajar acompañado por un adulto.

Por lo tanto tendrá un atributo: acompanado_por_adulto, que será booleano. 

#### Vuelo
Crear una clase Vuelo con: codigo, origen, destino, capacidad, lista de pasajeros
El código del vuelo será inmutable.
Debe tener los siguientes métodos:

- agregar_pasajero(pasajero)
- eliminar_pasajero(id_pasajero)
- hay_lugar()
- cantidad_pasajeros()
- calcular_recaudacion()

###### agregar_pasajero()
Debe agregar al pasajero solamente si:
    - Hay lugar disponible.
    - El pasajero puede abordar.

Debe devolver: True si fue agregado correctamente y False si no pudo ser agregado.

###### hay_lugar()
Debe devolver True si todavía hay lugares disponibles.

###### calcular_recaudacion()
Debe calcular cuánto dinero recaudó el vuelo utilizando:

pasajero.calcular_precio_vuelo()

No debe comprobar qué tipo de pasajero es.



```python
## RESOLUCION
from abc import ABC, abstractmethod

class Pasajero(ABC):
   
    def __init__(self, id, nombre, edad, equipaje_kg):
        self.__id = id
        self.__nombre = nombre
        self.__edad = edad
        self.__equipaje_kg = equipaje_kg

    ## ESCRIBIR GETTERS y SETTERS
    def get_id(self):
        return self.__id

    def get_nombre(self):
        return self.__nombre

    def get_equipaje(self):
        return self.__equipaje_kg
        
    def set_nombre(self, nombre):
        self.__nombre = nombre

    def es_mayor_de_edad(self):
        return self.__edad >=18

    def obtener_informacion():
        print(f" {self.__nombre} - {self.__edad} - {self.__equipaje_kg}")
    
    @abstractmethod
    def calcular_precio_vuelo(self):
        pass

    @abstractmethod
    def puede_abordar(self):
        pass


class PasajeroTurista(Pasajero):

    def calcular_precio_vuelo(self):
        eq_extra = max(0, self.get_equipaje() - 15)
        return((eq_extra)*2000) + 100000

    def puede_abordar(self):
        return (self.es_mayor_de_edad() and self.get_equipaje<=25)
    
    def obtener_informacion(self):
        print("SOY TURISTA")
        super().obtener_informacion()
        
    

class PasajeroEjecutivo(Pasajero):
    def __init__(self, id, nombre, edad, equipaje_kg, vuelos_realizados):
        super().__init__(id, nombre, edad, equipaje_kg)
        self.__vuelos_realizados = vuelos_realizados

        ## Crear Getters y Setters

    def calcular_precio_vuelo(self):
        eq_extra = max(0, self.get_equipaje() - 30)
        precio = ((eq_extra)*3000) + 200000
        if self.__vuelos_realizados > 5:
            precio = precio - (precio * 0.1)

        return precio


    def puede_abordar(self):
        return (self.es_mayor_de_edad() and self.get_equipaje<=40)
        

    def obtener_informacion(self):
        print("SOY EJECUTIVO")
        super().obtener_informacion()
        print(f"{self.__vuelos_realizados}")

          
class PasajeroMenor(Pasajero):

    def __init__(self, id, nombre, edad, equipaje_kg, acompanado_adulto):
        super().__init__(id, nombre, edad, equipaje_kg)
        self.__acompanado_adulto = acompanado_adulto

    ## Crear Getters y Setters
    
    def calcular_precio_vuelo(self):
        eq_extra = max(0, self.get_equipaje() - 10)

        return((eq_extra)*1500) + 50000
         
     
    def puede_abordar(self):
        return (not self.es_mayor_de_edad() and self.__acompanado_adulto)


    def obtener_informacion(self):
        print("SOY MENOR")
        super().obtener_informacion()
        print(f"{self.__acompanado_adulto}")


import unittest

class TestPasajeroTurista(unittest.TestCase):

    def test_es_mayor_de_edad(self):
        turista = PasajeroTurista("001", "Juan", 30, 10)
        self.assertTrue(turista.es_mayor_de_edad())

unittest.main(defaultTest='TestPasajeroTurista', argv=[''], verbosity=2, exit=False) 
```
