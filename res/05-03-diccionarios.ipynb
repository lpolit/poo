## Carrito

Escribir un programa que cree un diccionario simulando un carrito de compras. El programa debe preguntar el artículo y su precio y añadir el par al diccionario, hasta que el usuario decida terminar. Después se debe mostrar por pantalla la lista de la compra y el costo total, con el siguiente formato

* Lista de la compra
  * Artículo 1 Precio
  * Artículo 2 Precio
  * Artículo 3 Precio
  * ...        ...
  * Total Coste

```python
carrito = {}
art = input('Artículo: ')
carrito[art] = float(input('Precio $: '))
opc = input('Continua (S/N): ')
while opc == 'S':
    art = input('Artículo: ')
    carrito[art] = float(input('Precio $: '))
    opc = input('Continua (S/N): ')
print('Lista de la compra')
total = 0
for art, precio in carrito.items():
    print(f'{art} {precio}')
    total += precio
print(f'Total {total}')
```

## Traductor

Escribir un programa que cree un diccionario de traducción español-inglés. El usuario introducirá las palabras en español e inglés separadas por dos puntos, y cada par \<palabra>:<traducción> separados por comas. El programa debe crear un diccionario con las palabras y sus traducciones. Después pedirá una frase en español y utilizará el diccionario para traducirla palabra a palabra. Si una palabra no está en el diccionario debe dejarla sin traducir

```python
diccionario = {}
palabras_usuario = input('Palabras (español:ingles,...): ')
palabras = palabras_usuario.split(',')
for i in range(0, len(palabras)):
    traduccion = palabras[i].split(':')
    diccionario[traduccion[0]] = traduccion[1]
frase = input('Frase: ')
frase_parse = frase.split()
frase_traducida = ''
for i in range(0, len(frase_parse)):
    if frase_parse[i] not in diccionario:
        frase_traducida += f'{frase_parse[i]} '
    else:
        frase_traducida += f'{diccionario.get(frase_parse[i])} '
print(frase_traducida)
```

## Clientes CRUD

Escribir un programa que permita gestionar la base de datos de clientes de una empresa. Los clientes se guardarán en un diccionario en el que la clave de cada cliente será su DNI, y el valor será otro diccionario con los datos del cliente (nombre, dirección, teléfono, correo electrónico, preferente), donde preferente tendrá el valor True si se trata de un cliente preferente. El programa debe preguntar al usuario por una opción del siguiente menú: (1) Añadir cliente, (2) Eliminar cliente, (3) Mostrar cliente, (4) Listar todos los clientes, (5) Listar clientes preferentes, (6) Terminar. En función de la opción elegida el programa tendrá que hacer lo siguiente:

1. Preguntar los datos del cliente, crear un diccionario con los datos y añadirlo a la base de datos.
2. Preguntar por el DNI del cliente y eliminar sus datos de la base de datos.
3. Preguntar por el DNI del cliente y mostrar sus datos.
4. Mostrar lista de todos los clientes de la base datos con su DNI y nombre.
5. Mostrar la lista de clientes preferentes de la base de datos con su DNI y nombre.
6. Terminar el programa

```python
clientes = {'123': {'nom': 'Pablo', 'dir': 'abc', 'tel': '123', 'mail': '', 'pref': True}, 
            '456': {'nom': 'Ana', 'dir': 'abc', 'tel': '123', 'mail': '', 'pref': False}, 
            '789': {'nom': 'Maria', 'dir': 'abc', 'tel': '123', 'mail': '', 'pref': True}}

def menu():
    opc = 0
    while (opc != 6):
        if opc == 1:
            agregar()
        if opc == 2:
            eliminar()
        if opc == 3:
            consultar()
        if opc == 4:
            listar()
        if opc == 5:
            listar_pref()
        if opc != 0:
            input()
        print()
        print('--------MENU---------')
        print('1. Agregar cliente')
        print('2. Eliminar cliente')
        print('3. Consultar cliente')
        print('4. Listado clientes')
        print('5. Clientes preferentes')
        print('6. Salir')
        print()
        opc = int(input('?- '))

def agregar():
    dni = input('DNI: ')
    datos = {}
    datos['nom'] = input('Nombre: ')
    datos['dir'] = input('Dirección: ')
    datos['tel'] = input('Tel. :')
    datos['mail'] = input('Email: ')
    datos['pref'] = True if input('Preferente (S/N): ') == 'S' else False
    clientes[dni] = datos

def eliminar():
    dni = input('DNI: ')
    if dni in clientes.keys():
        clientes.pop(dni)
    else:
        print('Cliente no encontrado...')

def consultar():
    dni = input('DNI: ')
    if dni in clientes.keys():
        print(datos(dni, clientes[dni]))
    else:
        print('Cliente no encontrado...')

def datos(dni, cliente):
    return f"{dni}, {cliente['nom']}, {cliente['dir']}, {cliente['tel']}, {cliente['mail']}, {cliente['pref']}"

def listar_pref():
    for elem in clientes.items():
        if elem[1]['pref']:
            print(datos(elem[0], elem[1]))

def listar():
    for elem in clientes.items():
        print(datos(elem[0], elem[1]))
                
menu()
```

### Sistema de inventario
Una tienda necesita administrar sus productos.

Cada producto tiene un código, un nombre y un precio. 

No se debe permitir modificar directamente sus atributos desde fuera de la clase.

Se pide implementar:
1. Una clase Producto con los atributos encapsulados.
2. Una clase Inventario que almacene los productos en un diccionario, utilizando el código como clave y el objeto Producto como valor.
3. Un método para agregar un producto. No se deben permitir códigos duplicados.
4. Un método para buscar un producto por su código.
5. Un método para eliminar un producto.
6. Un método que muestre todos los productos con sus precios.
7. Un método que calcule el valor total del inventario.
Condición: el inventario debe almacenar objetos de la clase Producto, no solamente sus nombres y precios.

```python
## 1
class Producto:
    def __init__(self, codigo, nombre, precio):
        self.__codigo = codigo
        self.__nombre = nombre
        self.__precio = precio

    @property
    def codigo(self):
        return self.__codigo

    @property
    def nombre(self):
        return self.__nombre

    @property
    def precio(self):
        return self.__precio

    def __str__(self):
        return f"{self.__codigo} - {self.__nombre}: ${self.__precio}"

## 2
class Inventario:
    def __init__(self):
        self.__productos = {}


    ## 3
    def agregar_producto(self, prod):
        #if prod.codigo in self.__productos:
        #    return False

        if self.buscar_producto(prod.codigo) is None:
            self.__productos[prod.codigo] = prod
            return True

        return False

    ## 4
    def buscar_producto(self, codigo):
        return self.__productos.get(codigo)


    ## 5
    def eliminar_producto(self, codigo):
        self.__productos.pop(codigo)

    ## 6
    def mostrar_productos(self):
        for prod in self.__productos.values():
            print(prod)
    
    
    def calcular_valor_total(self):
        total = 0
        for prod in self.__productos.values():
            total += prod.precio 
        return total

        # return sum(prod.precio for prod in self.__productos.values())

inv = Inventario()

p1 = Producto("001", "Teclado", 2500)
p2 = Producto("002", "Mouse", 1500)
p3 = Producto("003", "Monitor", 10000)

inv.agregar_producto(p1)
inv.agregar_producto(p2)
inv.agregar_producto(p3)

inv.mostrar_productos()

print("VALOR TOTAL")
print("Valor total: ",inv.calcular_valor_total())

print("BUSCAR PRODUCTO")
print(inv.buscar_producto("001"))

print("ELIMINAR PRODUCTO")
inv.eliminar_producto("003")

print("INVENTARIO ACTUALIZADO")
inv.mostrar_productos()

print("VALOR TOTAL")
print("Valor total: ",inv.calcular_valor_total())

    
```

### Catálogo de productos con herencia y polimorfismo
Una tienda vende productos físicos y productos digitales.

Todos los productos tienen un código, un nombre y un precio base.

Los productos físicos tienen un costo de envío de $5.000, mientras que los productos digitales no tienen costo de envío.

Se pide implementar:
1. Una clase Producto que represente los datos comunes.
2. Una clase ProductoFisico que herede de Producto.
3. Una clase ProductoDigital que herede de Producto.
4. Un método calcular_precio_final() que devuelva el precio base más el costo de envío cuando corresponda.
5. Una clase Catalogo que almacene los productos en un diccionario.
6. Un método que permita agregar productos sin aceptar códigos duplicados.
7. Un método que muestre el precio final de todos los productos.
8. Un método que calcule el importe total de todos los productos del catálogo.


```python
## 1
class Producto:
    def __init__(self, codigo, nombre, precio_base):
        self.__codigo = codigo
        self.__nombre = nombre
        self.__precio_base = precio_base

    @property
    def codigo(self):
        return self.__codigo

    @property
    def nombre(self):
        return self.__nombre

    @property
    def precio_base(self):
        return self.__precio_base
    
    def calcular_precio_final(self):
        return self.__precio_base
        
    def __str__(self):
        return f"{self.__codigo} - {self.__nombre}: ${self.__precio_base}"

class ProductoDigital(Producto):
    pass

class ProductoFisico(Producto):
    COSTO_ENVIO = 5000
    
    def calcular_precio_final(self):
        return self.precio_base + self.COSTO_ENVIO


class Catalogo:
    def __init__(self):
        self.__productos = {}
    
    ##Un método que permita agregar productos sin aceptar códigos duplicados.
    def agregar_prod(self, prod):
        if prod.codigo in self.__productos:
            return False
    
        self.__productos[prod.codigo] = prod
        return True

    #Un método que muestre el precio final de todos los productos.
    def mostrar_precios(self):        
        for prod in self.__productos.values():
            print(prod.codigo, ":", prod.calcular_precio_final())

    # 
    def calcular_importe_base(self):
         return sum(prod.precio_base
            for prod in self.__productos.values()
        )

    
    #Un método que calcule el importe total de todos los productos del catálogo.
    def calcular_importe_total(self):
        return sum( prod.calcular_precio_final()
            for prod in self.__productos.values()
        )


catalogo = Catalogo()

p1 = ProductoFisico("001", "Teclado", 2500)
p2 = ProductoFisico("002", "Mouse", 1500)
p3 = ProductoDigital("003", "Curso Python", 30000)
p4 = ProductoDigital("004", "Libro electronico", 40000)

catalogo.agregar_prod(p1)
catalogo.agregar_prod(p2)
catalogo.agregar_prod(p3)
catalogo.agregar_prod(p4)

catalogo.mostrar_precios()
print("SUMA DE IMPORTES BASE")
print(catalogo.calcular_importe_base())
print("SUMA DE IMPORTES TOTAL")
print(catalogo.calcular_importe_total())
```
