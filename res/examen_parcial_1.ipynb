## Examen Parcial 1
1) Una empresa de envíos necesita digitalizar la gestión de los paquetes que ofrece a
sus clientes. Cada paquete posee un destino, un precio y un peso (por defecto 0 kg). 

No se permite crear instancias de la clase Paquete.

Existen diferentes tipos de paquetes:

- Los PaquetesNacionales deben almacenar la provincia de destino.
- Los PaquetesInternacionales deben almacenar el país de destino.
Cada tipo de paquete debe implementar un método que permita mostrar por pantalla la
información correspondiente al paquete, independientemente del tipo de paquete que sea.

La superclase debe llevar un registro de la cantidad de objetos instanciados y proporcionar
un método que permita retornar dicha cantidad.

2) Además, la empresa necesita calcular el costo total de enviar dos paquetes juntos.
Implemente un método (costo_total) que, dado dos paquetes y un monto máximo disponible,
indique si es posible enviar ambos paquetes sin superar dicho presupuesto.

3) Realizar un test que permita verificar que el método costo_total funciona correctamente.


Responder V o F:
- La sobrescritura permite que una subclase cambie la implementación heredada de un método.
- Un atributo de clase puede ser compartido por todas las instancias.
- Una clase abstracta debe poder instanciarse directamente para que sus métodos abstractos funcionen.
- Una clase puede tener métodos que no modifiquen ningún atributo del objeto.
- La composición representa una relación en la que un objeto puede contener o utilizar otros objetos.


```python
from abc import ABC, abstractmethod

class Paquete(ABC):
    cantidad = 0
    
    def __init__(self, precio, peso=0):
        self.precio = precio
        self.peso = peso
        Paquete.cantidad+=1

    @abstractmethod
    def mostrar_informacion(self):
        print("Precio:", self.precio)
        print("Peso:", self.peso, "Kg")

    @classmethod
    def obtener_cantidad(cls):
        return Paquete.cantidad

class PaqueteNacional(Paquete):
    def __init__(self, precio, provincia, peso=0):
        super().__init__(precio, peso)
        self.provincia = provincia

    def mostrar_informacion(self):
        print("Paquete Nacional")
        print("Provincia:", self.provincia)
        super().mostrar_informacion()

        
class PaqueteInternacional(Paquete):
    def __init__(self, precio, pais, peso=0):
        super().__init__(precio, peso)
        self.pais = pais

    def mostrar_informacion(self):
        print("Paquete Internacion")
        print("Pais:", self.pais)
        super().mostrar_informacion()

## Punto 2
def costo_total(paquete1, paquete2, monto_maximo):
    total = paquete1.precio + paquete2.precio
    if total <= monto_maximo:
        return total
    return -1


## Punto 3
import unittest

class TestCostoTotal(unittest.TestCase):

    def test_presupuesto_suficiente(self):
        p1 = PaqueteNacional(5000, "Cordoba",5)
        p2 = PaqueteInternacional(15000, "Brasil",10)

        self.assertTrue(costo_total(p1, p2, 50000)>=0)

    def test_presupuesto_insuficiente(self):
        p1 = PaqueteNacional(5000, "Cordoba",5)
        p2 = PaqueteInternacional(15000, "Brasil",10)

        self.assertFalse(costo_total(p1, p2, 1000)>=0)

    def test_presupuesto_exacto(self):
        p1 = PaqueteNacional(5000, "Cordoba",5)
        p2 = PaqueteInternacional(15000, "Brasil",10)

        self.assertTrue(costo_total(p1, p2, 20000)>=0)
        
      
unittest.main(defaultTest="TestCostoTotal", argv=[''],verbosity=2,exit=False)

        
```
