# Sistema de gestión de materiales de una biblioteca

Una biblioteca necesita desarrollar un sistema para administrar los materiales que ofrece en préstamo. Actualmente, la biblioteca presta dos tipos de materiales: libros y revistas. Cada préstamo tiene un costo asociado, que depende del tipo de material y de sus características.

Se solicita desarrollar una aplicación que permita registrar materiales, calcular sus costos de préstamo, almacenar la información en un archivo y recuperarla posteriormente.

### 1. Implementación de clases

Se deberán implementar tres clases: `Material`, `Libro` y `Revista`.

**Clase Material**

Representa un material genérico de la biblioteca. Deberá contar con los siguientes atributos:

- `codigo`: identificador numérico del material.
- `titulo`: título del material.
- `dias_prestamo`: cantidad de días que dura el préstamo.

Se deberán implementar los siguientes métodos:

- Método `calcular_costo()` que devuelva el costo del préstamo.
- Método `mostrar_informacion()` que muestre los datos del material.

Los atributos deberán estar encapsulados. Además, el constructor deberá validar que el código sea positivo, que el título no esté vacío y que la cantidad de días de préstamo sea mayor a cero.

**Clase Libro**

Hereda de la clase `Material`.

Deberá incorporar un atributo adicional:

- `valor_dia`: costo correspondiente a cada día de préstamo.

Deberá sobrescribir el método `calcular_costo()`, que deberá calcular el costo total del préstamo multiplicando la cantidad de días por el valor diario.

También deberá sobrescribir el método `mostrar_informacion()` para mostrar los datos específicos del libro.

El valor diario deberá ser mayor a cero.

**Clase Revista**

Hereda de la clase `Material`.

Deberá incorporar un atributo adicional:

- `costo_fijo`: importe fijo correspondiente al préstamo de la revista.

Deberá sobrescribir el método `calcular_costo()`, que deberá devolver el costo fijo establecido.

También deberá sobrescribir el método `mostrar_informacion()` para mostrar los datos específicos de la revista.

El costo fijo deberá ser mayor a cero.

### 2. Manejo de excepciones

Se deberán implementar mecanismos de tratamiento de excepciones para contemplar las siguientes situaciones:

- Si se intenta crear un material con un código inválido, un título vacío o una cantidad de días incorrecta, se deberá generar una excepción `ValueError`.
- Si se intenta crear un libro con un valor diario menor o igual a cero, se deberá generar una excepción `ValueError`.
- Si se intenta crear una revista con un costo fijo menor o igual a cero, se deberá generar una excepción `ValueError`.
- Si se intenta recuperar información desde un archivo que no existe, se deberá capturar la excepción `FileNotFoundError` e informar al usuario.
- Si durante la lectura del archivo se encuentran datos incorrectos o registros con un formato inválido, se deberán capturar las excepciones correspondientes e informar el problema.

Las excepciones deberán tratarse de manera que el programa no finalice inesperadamente ante un error.

### 3. Lectura y escritura de archivos

La información de los materiales deberá almacenarse en un archivo de texto llamado `materiales.csv`, utilizando el carácter punto y coma (`;`) como separador de campos.

Cada línea del archivo deberá contener los siguientes datos:

`tipo;codigo;titulo;dias_prestamo;dato_adicional`

Donde:

- `tipo` identifica si el material es un libro o una revista.
- `codigo` representa el código del material.
- `titulo` representa su título.
- `dias_prestamo` indica la cantidad de días del préstamo.
- `dato_adicional` representa el valor diario en el caso de los libros o el costo fijo en el caso de las revistas.

A modo de ejemplo, el archivo podría contener los siguientes registros:

```text
LIBRO;1;El Principito;5;200
REVISTA;2;National Geographic;3;500
LIBRO;3;Don Quijote;7;150
REVISTA;4;Muy Interesante;4;350
```

Se deberán implementar las siguientes funciones:

**a. guardar_materiales(materiales)**

Recibirá una lista de objetos de tipo `Material` y almacenará sus datos en el archivo `materiales.csv`.

Se deberá identificar el tipo de cada objeto para almacenar correctamente la información específica de cada subclase.

**b. recuperar_materiales()**

Leerá el archivo `materiales.csv` y construirá una lista de objetos a partir de los registros almacenados.

Se deberá identificar el tipo de material de cada registro y crear un objeto de la clase correspondiente (`Libro` o `Revista`).

La función deberá devolver la lista de objetos recuperados.

### 4. Implementación de un decorador 

Se deberá implementar un decorador denominado `registrar_operacion`.

El decorador deberá mostrar por pantalla un mensaje antes de ejecutar la función decorada y otro mensaje una vez finalizada su ejecución.

Por ejemplo, al ejecutar la función `guardar_materiales()`, se deberá mostrar:

```text
Iniciando operación: guardar_materiales
Operación finalizada: guardar_materiales
```

El decorador deberá aplicarse a las funciones `guardar_materiales()` y `recuperar_materiales()`.

### 5. Programa principal

Implementar un programa principal que permita realizar las siguientes operaciones:

1. Crear al menos dos objetos de la clase `Libro` y dos objetos de la clase `Revista`.
2. Almacenar todos los objetos creados en una única lista.
3. Recorrer la lista y mostrar la información de cada material, incluyendo el costo de su préstamo.
4. Guardar todos los materiales en el archivo `materiales.csv`.
5. Recuperar los materiales almacenados en el archivo.
6. Recorrer la lista recuperada y mostrar nuevamente la información de cada material.

El programa principal deberá contemplar el tratamiento de las excepciones que puedan producirse durante la creación de objetos y durante las operaciones de lectura y escritura de archivos.
