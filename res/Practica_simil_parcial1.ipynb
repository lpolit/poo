## Ejercicio — Gestión de publicaciones

Una plataforma permite publicar distintos tipos de contenido. 
Actualmente existen publicaciones de texto y publicaciones de video.

Cada publicación tiene un título, un autor y una cantidad de visualizaciones. 
La plataforma permite mostrar información de una publicación y calcular los puntos obtenidos por ella.

Las publicaciones de texto y de video calculan sus puntos de manera diferente:

- Una publicación de texto obtiene 2 puntos por cada visualización.
- Una publicación de video obtiene 5 puntos por cada visualización.

Además, una publicación de video posee una calidad, que puede ser HD, FULL_HD o 4K.

Se pide desarrollar una solución orientada a objetos que permita representar esta situación.

a)Implementar las clases definidas, 
- Contemplar la relación que existe entre ellas
- Qué atributos debe tener cada una. La información que no deba ser modificada directamente desde fuera de la clase deberá estar encapsulada.
- La solución debe contemplar la utilización de un enumerado para representar la calidad de los videos.

b) Implementar una función llamada mostrar_puntos(publicacion) que reciba una publicación y muestre la cantidad de puntos obtenidos.

La función debe funcionar correctamente tanto si recibe una publicación de texto como una publicación de video, sin consultar explícitamente de qué tipo de publicación se trata.
